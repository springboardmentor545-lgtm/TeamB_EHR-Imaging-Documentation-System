```python
import os
import cv2
import numpy as np
from sklearn.model_selection import train_test_split

# Input folder (original Pneumonia images)
input_folder =r"C:\Users\gbhoo\OneDrive\Attachments\images_cleaned\chest_xray\Pneumonia" 
# Output folders
train_folder = r"processed_pneumonia/train"
test_folder = r"processed_pneumonia/test"

os.makedirs(train_folder, exist_ok=True)
os.makedirs(test_folder, exist_ok=True)

images = []
count = 0

# Read and preprocess 30 images
for file in os.listdir(input_folder):
    if file.lower().endswith((".jpg", ".jpeg", ".png")):
        img_path = os.path.join(input_folder, file)

        # Read as grayscale
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)

        # Resize
        img = cv2.resize(img, (256, 256))

        # Normalize
        img = img / 255.0

        images.append(img)
        count += 1

        if count == 30:  # only take 30 images
            break

images = np.array(images)

# Train/Test split (80/20)
X_train, X_test = train_test_split(images, test_size=0.2, random_state=42)

# Save images into respective folders
for i, img in enumerate(X_train):
    save_path = os.path.join(train_folder, f"pneumonia_train_{i}.png")
    cv2.imwrite(save_path, (img * 255).astype(np.uint8))

for i, img in enumerate(X_test):
    save_path = os.path.join(test_folder, f"pneumonia_test_{i}.png")
    cv2.imwrite(save_path, (img * 255).astype(np.uint8))

print("Preprocessing + Train/Test Split Complete!")
print("Train images:", len(X_train))
print("Test images:", len(X_test))
```
